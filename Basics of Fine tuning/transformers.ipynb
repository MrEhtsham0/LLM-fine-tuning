# **Fine Tuning of Tiny LLama**

Full Fine-Tuning (FFT) is expensive because it updates all model parameters, leading to high computational cost (massive VRAM for gradients/optimizers) and high storage cost (a full model copy per task).

LoRA solves this by leveraging the insight that weight updates have low intrinsic rank. Instead of updating $\Delta W$ directly, LoRA freezes the base model and injects small, trainable low-rank matrices ($A$ and $B$) to approximate $\Delta W = BA$, drastically reducing memory and storage requirements.

Sets the padding token to be the same as the end-of-sequence token. This is necessary because some models don't have a defined padding token by default. Using EOS token ensures consistent token IDs and prevents issues during batching where sequences need to be padded to the same length.

## **Creates LoRA configuration**:

1. r=16: Rank of LoRA matrices (the r from theoretical explanation). Lower = fewer parameters, higher = more expressive power

2. lora_alpha=32: Scaling factor. The actual scaling is alpha/r = 32/16 = 2. This controls how much the LoRA adapter influences the output

3. target_modules: Which attention layers to apply LoRA to (query, value, key, output projections). Common practice to target attention layers only

4. lora_dropout=0.05: Dropout rate for LoRA layers (regularization)

5. bias="none": Don't train bias terms (only train LoRA parameters)

6. task_type="CAUSAL_LM": Task type (causal language modeling, like GPT)

```python
print("✅ Loading model and tokenizer...")
tokenizer.pad_token = tokenizer.eos_token


# -------------------------------
# 2. Load and Preprocess Dataset (resume Q&A CSV)
# -------------------------------

csv_path = Path("resume_dataset.csv")
if not csv_path.is_file():
    raise FileNotFoundError(
        "Create resume_dataset.csv first (run the cell above that builds the DataFrame)."
    )

dataset = load_dataset("csv", data_files=str(csv_path))["train"]


def format_chat_batch(batch):
    texts = []
    for question, answer in zip(batch["question"], batch["answer"]):
        messages = [
            {"role": "user", "content": question},
            {"role": "assistant", "content": answer},
        ]
        texts.append(
            tokenizer.apply_chat_template(
                messages, tokenize=False, add_generation_prompt=False
            )
        )
    enc = tokenizer(texts, truncation=True, padding="max_length", max_length=512)
    labels = []
    for ids, mask in zip(enc["input_ids"], enc["attention_mask"]):
        labels.append([tid if m else -100 for tid, m in zip(ids, mask)])
    enc["labels"] = labels
    return enc


print("🔁 Tokenizing resume dataset...")

tokenized_dataset = dataset.map(
    format_chat_batch,
    batched=True,
    batch_size=8,
    num_proc=1,
    remove_columns=dataset.column_names,
    desc="Tokenizing resume Q&A",
)

# -------------------------------
# 3. Apply LoRA
# -------------------------------

print("✅ Applying LoRA configuration...")

lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    target_modules=["q_proj", "v_proj", "k_proj", "o_proj"],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
)

model = get_peft_model(model, lora_config)

model.print_trainable_parameters()

# -------------------------------
# 4. Training Setup
# -------------------------------

use_fp16 = torch.cuda.is_available()

training_args = TrainingArguments(
    output_dir="./tinyllama-lora-resume",
    per_device_train_batch_size=1,
    gradient_accumulation_steps=1,
    learning_rate=1e-4,
    num_train_epochs=30,
    fp16=use_fp16,
    logging_steps=1,
    save_steps=50,
    save_total_limit=2,
    report_to="none",
)

trainer = Trainer(model=model, args=training_args, train_dataset=tokenized_dataset)

# -------------------------------
# 5. Start Fine-Tuning
# -------------------------------

print("🚀 Starting fine-tuning...")

trainer.train()

# -------------------------------
# 6. Save LoRA Adapter
# -------------------------------

print("💾 Saving the fine-tuned LoRA adapter...")

trainer.save_model("./tinyllama-lora-resume-final")
tokenizer.save_pretrained("./tinyllama-lora-resume-final")

print("✅ Training & saving complete!")
```

## **Why QLoRA Works Better for Consumer GPUs**
4-bit Base Model: The frozen base weights take 4× less memory than fp16

NF4 Quantization: Normal Float 4 preserves more precision than standard 4-bit

Double Quantization: Further reduces memory by quantizing the quantization constants

Paged Optimizer: Uses CPU RAM for optimizer states when GPU memory is full

## **When to Choose QLoRA vs LoRA**
### **Choose QLoRA when:**

You have limited GPU memory (<8GB)

You want to fine-tune larger models (13B, 70B)

You want to use larger batch sizes

### **Choose LoRA when:**

You have sufficient GPU memory (10GB+ for 7B models)

Training speed is critical (QLoRA has ~10-20% overhead)

You need maximum precision (rarely matters for fine-tuning)

```python
#!/usr/bin/env python3
"""
QLoRA Fine-tuning Script for Resume Q&A Dataset
This script fine-tunes TinyLlama on a resume Q&A dataset using QLoRA (4-bit quantization + LoRA)
"""

import torch
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    Trainer,
    TrainingArguments,
    BitsAndBytesConfig,
)
from peft import LoraConfig, get_peft_model
from datasets import load_dataset
from pathlib import Path
import pandas as pd
import os

# -------------------------------
# Configuration
# -------------------------------

MODEL_NAME = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
OUTPUT_DIR = "./tinyllama-qlora-resume"
FINAL_OUTPUT_DIR = "./tinyllama-qlora-resume-final"
CSV_PATH = "resume_dataset.csv"
MAX_LENGTH = 512
LORA_R = 16
LORA_ALPHA = 32
LORA_DROPOUT = 0.05
LEARNING_RATE = 2e-4
NUM_EPOCHS = 30
BATCH_SIZE = 4
GRADIENT_ACCUMULATION_STEPS = 1

# -------------------------------
# 1. Create Sample Dataset (if not exists)
# -------------------------------


def create_sample_dataset():
    """Create a sample resume Q&A dataset if it doesn't exist"""
    if Path(CSV_PATH).is_file():
        print(f"✅ Dataset {CSV_PATH} already exists")
        return

    print(f"📝 Creating sample dataset: {CSV_PATH}")

    sample_data = {
        "question": [
            "What are your key strengths?",
            "Tell me about your work experience.",
            "What technologies are you proficient in?",
            "Describe a challenging project you worked on.",
            "Why do you want this position?",
            "What is your educational background?",
            "How do you handle tight deadlines?",
            "What are your career goals?",
        ],
        "answer": [
            "I have strong problem-solving skills, excellent communication abilities, and a proven track record of delivering high-quality software solutions. I'm adaptable and quick to learn new technologies.",
            "I have 5 years of experience in software development, working primarily with Python and cloud technologies. At my previous role at TechCorp, I led a team of 3 developers to build a microservices architecture that reduced deployment time by 40%.",
            "I'm proficient in Python, JavaScript, React, Node.js, and cloud platforms like AWS and Azure. I also have experience with databases including PostgreSQL and MongoDB, and I'm comfortable with DevOps tools like Docker and Kubernetes.",
            "I developed a real-time analytics dashboard that processed over 1 million events per day. The challenge was optimizing performance while maintaining accuracy. I implemented caching strategies and optimized database queries, resulting in a 60% reduction in load time.",
            "I'm passionate about this position because it aligns perfectly with my skills in full-stack development and my interest in building scalable applications. I admire your company's commitment to innovation and would love to contribute to your upcoming projects.",
            "I hold a Master's degree in Computer Science from Stanford University and a Bachelor's in Software Engineering from MIT. I've also completed several certifications in cloud architecture and machine learning.",
            "I prioritize tasks based on urgency and impact, break down large projects into manageable milestones, and maintain clear communication with stakeholders. I use tools like Jira and Trello to track progress and ensure timely delivery.",
            "My short-term goal is to grow as a technical leader while delivering impactful solutions. Long-term, I aspire to become an architect where I can shape technical strategy and mentor junior developers.",
        ],
    }

    # Create more variations by duplicating and modifying slightly
    df = pd.DataFrame(sample_data)

    # Duplicate to create more examples (for better training)
    df = pd.concat([df] * 5, ignore_index=True)

    # Save to CSV
    df.to_csv(CSV_PATH, index=False)
    print(f"✅ Created {len(df)} training examples in {CSV_PATH}")


# -------------------------------
# 2. Setup QLoRA Quantization
# -------------------------------


def setup_quantization():
    """Configure 4-bit quantization for QLoRA"""
    bnb_config = BitsAndBytesConfig(
        load_in_4bit=True,  # Enable 4-bit loading
        bnb_4bit_quant_type="nf4",  # Normal Float 4 (optimal for weights)
        bnb_4bit_compute_dtype=torch.float16,  # Compute in fp16
        bnb_4bit_use_double_quant=True,  # Double quantization for memory savings
    )
    return bnb_config


# -------------------------------
# 3. Load Model with QLoRA
# -------------------------------


def load_model_and_tokenizer():
    """Load quantized model and tokenizer"""
    print("🔄 Loading model and tokenizer with QLoRA quantization...")

    # Setup quantization
    bnb_config = setup_quantization()

    # Load quantized model
    model = AutoModelForCausalLM.from_pretrained(
        MODEL_NAME,
        quantization_config=bnb_config,
        device_map="auto",  # Auto-distribute across available devices
        torch_dtype=torch.float16,
        trust_remote_code=True,
    )

    # Enable gradient checkpointing to save memory
    model.gradient_checkpointing_enable()

    # Load tokenizer
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
    tokenizer.pad_token = tokenizer.eos_token

    print("✅ Model and tokenizer loaded successfully")

    return model, tokenizer


# -------------------------------
# 4. Load and Preprocess Dataset
# -------------------------------


def load_and_preprocess_dataset(tokenizer):
    """Load CSV dataset and tokenize it"""
    print("📊 Loading dataset...")

    if not Path(CSV_PATH).is_file():
        raise FileNotFoundError(
            f"Dataset {CSV_PATH} not found. Run the script again to create a sample dataset."
        )

    # Load dataset
    dataset = load_dataset("csv", data_files=str(CSV_PATH))["train"]
    print(f"✅ Loaded {len(dataset)} examples")

    def format_chat_batch(batch):
        """Format Q&A pairs into chat template and tokenize"""
        texts = []
        for question, answer in zip(batch["question"], batch["answer"]):
            messages = [
                {"role": "user", "content": question},
                {"role": "assistant", "content": answer},
            ]
            texts.append(
                tokenizer.apply_chat_template(
                    messages, tokenize=False, add_generation_prompt=False
                )
            )

        # Tokenize
        enc = tokenizer(
            texts,
            truncation=True,
            padding="max_length",
            max_length=MAX_LENGTH,
            return_tensors="pt",
        )

        # Create labels (ignore padding tokens in loss)
        labels = []
        for ids, mask in zip(enc["input_ids"], enc["attention_mask"]):
            labels.append([tid if m else -100 for tid, m in zip(ids, mask)])

        enc["labels"] = labels
        return enc

    print("🔁 Tokenizing dataset...")

    # Tokenize dataset
    tokenized_dataset = dataset.map(
        format_chat_batch,
        batched=True,
        batch_size=8,
        num_proc=1,
        remove_columns=dataset.column_names,
        desc="Tokenizing",
    )

    print("✅ Dataset tokenization complete")

    return tokenized_dataset


# -------------------------------
# 5. Configure LoRA
# -------------------------------


def configure_lora(model):
    """Apply LoRA configuration to the quantized model"""
    print("🔧 Configuring LoRA...")

    lora_config = LoraConfig(
        r=LORA_R,  # Rank
        lora_alpha=LORA_ALPHA,  # Scaling factor
        target_modules=["q_proj", "v_proj", "k_proj", "o_proj"],  # Attention layers
        lora_dropout=LORA_DROPOUT,  # Dropout for regularization
        bias="none",  # Don't train bias terms
        task_type="CAUSAL_LM",  # Causal language modeling
    )

    # Wrap model with LoRA
    model = get_peft_model(model, lora_config)

    # Print trainable parameters
    model.print_trainable_parameters()

    return model


# -------------------------------
# 6. Setup Training Arguments
# -------------------------------


def setup_training_args():
    """Configure training arguments optimized for QLoRA"""

    training_args = TrainingArguments(
        output_dir=OUTPUT_DIR,
        per_device_train_batch_size=BATCH_SIZE,
        gradient_accumulation_steps=GRADIENT_ACCUMULATION_STEPS,
        learning_rate=LEARNING_RATE,
        num_train_epochs=NUM_EPOCHS,
        fp16=True,  # Mixed precision training
        logging_steps=1,  # Log every step
        save_steps=50,  # Save checkpoint every 50 steps
        save_total_limit=2,  # Keep only 2 checkpoints
        report_to="none",  # Disable external logging
        optim="paged_adamw_8bit",  # QLoRA's 8-bit optimizer
        gradient_checkpointing=True,  # Memory optimization
        gradient_checkpointing_kwargs={"use_reentrant": False},
        warmup_steps=10,  # Warmup steps
        lr_scheduler_type="cosine",  # Cosine learning rate schedule
        dataloader_num_workers=2,  # Data loading threads
    )

    return training_args


# -------------------------------
# 7. Training Function
# -------------------------------


def train_model(model, tokenized_dataset, training_args):
    """Execute the training process"""
    print("\n🚀 Starting QLoRA fine-tuning...")
    print("=" * 50)

    # Initialize trainer
    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_dataset,
    )

    # Start training
    trainer.train()

    print("\n✅ Training complete!")

    return trainer


# -------------------------------
# 8. Save Model
# -------------------------------


def save_model(trainer, tokenizer):
    """Save the fine-tuned QLoRA adapter"""
    print("\n💾 Saving the fine-tuned QLoRA adapter...")

    # Create output directory if it doesn't exist
    os.makedirs(FINAL_OUTPUT_DIR, exist_ok=True)

    # Save model and tokenizer
    trainer.save_model(FINAL_OUTPUT_DIR)
    tokenizer.save_pretrained(FINAL_OUTPUT_DIR)

    print(f"✅ Model saved to {FINAL_OUTPUT_DIR}")
    print(
        f"📁 Adapter size: ~{sum(os.path.getsize(os.path.join(FINAL_OUTPUT_DIR, f)) for f in os.listdir(FINAL_OUTPUT_DIR) if os.path.isfile(os.path.join(FINAL_OUTPUT_DIR, f))) / 1024 / 1024:.2f} MB"
    )


# -------------------------------
# 9. Test Function (Optional)
# -------------------------------


def test_model():
    """Test the fine-tuned model with sample prompts"""
    print("\n🧪 Testing the fine-tuned model...")

    try:
        # Load the fine-tuned model
        from peft import PeftModel

        # Load base model with quantization
        bnb_config = setup_quantization()
        base_model = AutoModelForCausalLM.from_pretrained(
            MODEL_NAME,
            quantization_config=bnb_config,
            device_map="auto",
            torch_dtype=torch.float16,
        )

        # Load LoRA adapter
        model = PeftModel.from_pretrained(base_model, FINAL_OUTPUT_DIR)
        tokenizer = AutoTokenizer.from_pretrained(FINAL_OUTPUT_DIR)

        # Test prompts
        test_prompts = [
            "What are your key strengths?",
            "Tell me about your work experience.",
            "What technologies are you proficient in?",
        ]

        for prompt in test_prompts:
            print(f"\n📝 Question: {prompt}")

            # Format prompt
            messages = [{"role": "user", "content": prompt}]
            formatted_prompt = tokenizer.apply_chat_template(
                messages, tokenize=False, add_generation_prompt=True
            )

            # Generate response
            inputs = tokenizer(formatted_prompt, return_tensors="pt").to("cuda")
            outputs = model.generate(
                **inputs,
                max_new_tokens=100,
                temperature=0.7,
                do_sample=True,
                pad_token_id=tokenizer.pad_token_id,
            )

            response = tokenizer.decode(outputs[0], skip_special_tokens=True)
            # Extract only the assistant's response
            if "<|assistant|>" in response:
                response = response.split("<|assistant|>")[-1].strip()

            print(f"💬 Answer: {response}\n")

    except Exception as e:
        print(f"⚠️ Testing failed: {e}")
        print("You can test the model manually after training.")


# -------------------------------
# 10. Main Execution
# -------------------------------


def main():
    """Main execution function"""
    print("=" * 50)
    print("🤗 QLoRA Fine-tuning for Resume Q&A")
    print("=" * 50)

    # Step 1: Create dataset if needed
    create_sample_dataset()

    # Step 2: Load model and tokenizer
    model, tokenizer = load_model_and_tokenizer()

    # Step 3: Load and preprocess dataset
    tokenized_dataset = load_and_preprocess_dataset(tokenizer)

    # Step 4: Configure LoRA
    model = configure_lora(model)

    # Step 5: Setup training arguments
    training_args = setup_training_args()

    # Step 6: Train the model
    trainer = train_model(model, tokenized_dataset, training_args)

    # Step 7: Save the model
    save_model(trainer, tokenizer)

    # Step 8: Test the model (optional)
    test_model()

    print("\n" + "=" * 50)
    print("✅ QLoRA Fine-tuning Complete!")
    print("=" * 50)
    print("\n📊 Training Summary:")
    print(f"   - Model: {MODEL_NAME}")
    print(f"   - LoRA Rank: {LORA_R}")
    print(f"   - Batch Size: {BATCH_SIZE}")
    print(f"   - Learning Rate: {LEARNING_RATE}")
    print(f"   - Epochs: {NUM_EPOCHS}")
    print(f"\n💾 Model saved to: {FINAL_OUTPUT_DIR}")
    print("\n🔧 To use the fine-tuned model:")
    print("   from peft import PeftModel")
    print("   from transformers import AutoModelForCausalLM, AutoTokenizer")
    print(
        "   model = PeftModel.from_pretrained(base_model, './tinyllama-qlora-resume-final')"
    )
    print(
        "   tokenizer = AutoTokenizer.from_pretrained('./tinyllama-qlora-resume-final')"
    )


# -------------------------------
# Run the script
# -------------------------------

if __name__ == "__main__":
    main()
```
